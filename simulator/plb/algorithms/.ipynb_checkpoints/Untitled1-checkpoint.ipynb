```python
from pathlib import Path
from datetime import datetime
import subprocess
import sys
import os
import time

workdir = Path.home() / "colcon_ws/src/RoboCraft/robocraft"

# seed 42는 이미 완료했으므로 43, 44만 추가 실행
seeds_to_run = [43]

env = os.environ.copy()
env["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

run_results = []

for run_number, seed in enumerate(seeds_to_run, start=2):

    # seed마다 별도의 결과 폴더 사용
    output_dir = (
        workdir
        / "runs"
        / f"{datetime.now().strftime('%Y%m%d_%H%M%S_%f')}_seed{seed}"
    )
    output_dir.mkdir(parents=True, exist_ok=False)

    cmd = [
        sys.executable, "-u", "control.py",

        "--stage", "control",
        "--data_type", "ngrip_fixed",

        "--model_path",
        "dump/dump_ngrip_fixed/"
        "out_25-Jun-2022-14:53:02.222126/"
        "net_epoch_19_iter_34.pth",

        "--shooting_batch_size", "50",
        "--control_algo", "predict",

        "--n_grips", "5",
        "--predict_horizon", "2",

        "--opt_algo", "GD",
        "--correction", "1",

        "--shape_type", "alphabet",
        "--goal_shape_name", "C",

        # 실행마다 다른 Random Shooting
        "--random_seed", str(seed),

        "--debug", "0",
        "--controlf", str(output_dir),
    ]

    log_path = output_dir / "console_log.txt"

    print("\n" + "=" * 70)
    print(f"Baseline {run_number}/3 실행 시작")
    print(f"Random seed: {seed}")
    print(f"결과 저장 위치: {output_dir}")
    print("=" * 70, flush=True)

    start_time = time.time()

    # 화면 출력과 로그 저장을 동시에 수행
    with log_path.open("w", encoding="utf-8") as log_file:

        with subprocess.Popen(
            cmd,
            cwd=str(workdir),
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True,
            env=env,
            bufsize=1,
        ) as process:

            for line in process.stdout:
                print(line, end="", flush=True)
                log_file.write(line)
                log_file.flush()

            return_code = process.wait()

    elapsed_minutes = (time.time() - start_time) / 60.0

    run_results.append({
        "baseline_run": run_number,
        "seed": seed,
        "return_code": return_code,
        "elapsed_minutes": elapsed_minutes,
        "output_dir": output_dir,
    })

    print(f"\nBaseline {run_number}/3 종료")
    print(f"Seed: {seed}")
    print(f"프로그램 종료 코드: {return_code}")
    print(f"실행 시간: {elapsed_minutes:.1f}분")
    print(f"결과 폴더: {output_dir}")

    # 오류가 발생하면 다음 실험은 실행하지 않음
    if return_code != 0:
        print("\n오류가 발생하여 다음 seed 실행을 중단합니다.")
        break


print("\n\n===== Baseline 실행 요약 =====")

# 기존 실험
print("Baseline 1/3: seed=42, 기존 실행 결과")

# 새 실험
for result in run_results:
    print(
        f"Baseline {result['baseline_run']}/3: "
        f"seed={result['seed']}, "
        f"종료 코드={result['return_code']}, "
        f"시간={result['elapsed_minutes']:.1f}분"
    )
    print(f"  저장 위치: {result['output_dir']}")
```

```python
from pathlib import Path
from PIL import Image as PILImage
import numpy as np
import hashlib

runs_root = Path(
    "/home/seohy/colcon_ws/src/RoboCraft/robocraft/runs"
)

# 최근 두 개의 결과
gifs = sorted(
    runs_root.rglob("best_anim.gif"),
    key=lambda p: p.stat().st_mtime,
    reverse=True
)[:2]

assert len(gifs) == 2, "비교할 GIF가 두 개 이상 필요합니다."

def file_hash(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

def final_frame(path):
    with PILImage.open(path) as im:
        im.seek(im.n_frames - 1)
        return np.asarray(im.convert("RGB"), dtype=np.int16)

frame_1 = final_frame(gifs[0])
frame_2 = final_frame(gifs[1])

print("최근 결과:", gifs[0])
print("이전 결과:", gifs[1])
print()
print("GIF 파일 전체가 완전히 동일:", file_hash(gifs[0]) == file_hash(gifs[1]))
print("최종 이미지가 픽셀 단위로 동일:", np.array_equal(frame_1, frame_2))
print("최종 이미지 평균 픽셀 차이:", np.abs(frame_1 - frame_2).mean())
print("최종 이미지 최대 픽셀 차이:", np.abs(frame_1 - frame_2).max())
```

```python
from pathlib import Path

repo_root = Path(
    "/home/seohy/colcon_ws/src/RoboCraft/robocraft"
)

keywords = [
    "random_seed",
    "manual_seed",
    "np.random.seed",
    "numpy.random.seed",
    "random.seed",
    "torch.seed",
    "set_seed",
]

matches = []

for path in repo_root.rglob("*.py"):
    # 실행 결과 폴더는 제외
    if "runs" in path.parts:
        continue

    try:
        lines = path.read_text(
            encoding="utf-8",
            errors="ignore"
        ).splitlines()
    except Exception:
        continue

    for line_number, line in enumerate(lines, start=1):
        if any(keyword in line for keyword in keywords):
            matches.append((path, line_number, line.strip()))

for path, line_number, line in matches:
    print(f"{path.relative_to(repo_root)}:{line_number}")
    print(f"    {line}")

print(f"\n총 {len(matches)}개 발견")
```
